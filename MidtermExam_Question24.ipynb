```python
def checker(s):

 if s.count('(') == s.count(')') and s.count('[') == s.count("]") and s.count('{') == s.count("}"):
  print(True)
 else:
  print(False)

print("({[]})")
checker("({[]})")

print("\n([)]")
checker("([)]")

print("\n((()))")
checker("((()))")

print("\n({})")
checker("({})")
```
