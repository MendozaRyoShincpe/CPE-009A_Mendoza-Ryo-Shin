Write a program that uses nested loops to print this pattern:

*#

**#

***#

****#

*****#

****#

***#

**#

*#

```python
for x in range(1,10):
  if x < 6:
    print(x * "*")
for y in range(4,0,-1):
  print(y * "*")
```
