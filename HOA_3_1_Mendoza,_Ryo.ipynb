**Distance Formula**
1. Make a program that would calculate the distance between two points given a list of coordinates. Use the distance
formula.
coorindates_list = [(1,1), (2,3)]

```python
import math

coordinates_list = [(1,1), (2,3)]

def calculate_distance(point1, point2):
    x1, y1 = point1
    x2, y2 = point2
    distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    return distance

# Calculate distance for the first two points in the list
p1 = coordinates_list[0]
p2 = coordinates_list[1]
dist = calculate_distance(p1, p2)
print(f"The distance between {p1} and {p2} is: {dist}")
```

**Simple Word Filter**

2. For a given string input, replace all the words “stupid” with an asterisk * equal to the length of the string. The new
string value should be displayed with the asterisks.

```python
text = "This is a stupid example with stupid words."
word_to_replace = "stupid"

replacement_string = '*' * len(word_to_replace)

filtered_text = text.replace(word_to_replace, replacement_string)

print(f"Original string: {text}")
print(f"Filtered string: {filtered_text}")
```

**Phonebook**

3. Create a simple phonebook program that can read from a list of dictionaries. The program should be able to display a
person’s name, contact, and address based from a user input which is the id of the record.

```python
phonebook_data = [
    {'id': 1, 'name': 'Alice Smith', 'contact': '123-456-7890', 'address': '123 Main St'},
    {'id': 2, 'name': 'Bob Johnson', 'contact': '098-765-4321', 'address': '456 Oak Ave'},
    {'id': 3, 'name': 'Charlie Brown', 'contact': '555-123-4567', 'address': '789 Pine Ln'}
]

def search_phonebook():
    try:
        user_id = int(input("Enter the ID of the contact you want to find: "))
        found = False
        for contact in phonebook_data:
            if contact['id'] == user_id:
                print(f"\nName: {contact['name']}")
                print(f"Contact: {contact['contact']}")
                print(f"Address: {contact['address']}")
                found = True
                break
        if not found:
            print(f"Contact with ID {user_id} not found.")
    except ValueError:
        print("Invalid input. Please enter a numerical ID.")

search_phonebook()
```

### 1. How do we display elements of lists, tuples, and strings?  
We display elements by using their index numbers, starting from 0. For example, `my_list[0]` shows the first item in a list, `my_tuple[1]` shows the second item in a tuple, and `my_string[2]` shows the third character in a string. It’s like picking something from a container by its position.

### 2. What is the difference between a list, tuple, string, and dictionary? Give possible use case for each.  
A list is a collection that can change, useful for things like a shopping list. A tuple is similar but cannot be changed, good for fixed data like coordinates. A string is a sequence of characters, perfect for storing text like names or sentences. A dictionary stores data in key-value pairs, useful when you want to look up information quickly, like a phonebook. Each has its own role depending on whether you need flexibility, fixed values, text, or fast lookups.
