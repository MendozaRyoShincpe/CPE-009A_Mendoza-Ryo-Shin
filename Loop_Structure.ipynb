Loop Statement

```python
# Example 1
week = ["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
for x in week:
  print(x)
```

```python
# Example 2
week = ["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
for x in week:
  print(x)
  if x == "Thursday":
    break
```

In Range Fnction

```python
for x in range(6):
  print(x)

for x in range(2,6):
  print("\n", x)
```

Nested Loops

```python
adjective = ["red","big","tasty"]
fruits = ["apple","banana","cherry"]

for x in adjective:
  for y in fruits:
    print(x,y)
```

While Loop

```python
i = 1
while i < 6:
  print(i)
  i += 1 #i = i+1 Assignment operator
```

Break Statement

```python
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i+=1
```

Else Statement

```python
i = 1
while i < 6:
  print(i)
i+=1
else:
  print("i is no longer less than 6")
```
